```python
#Use Case 1: Banking Eligibility Check
age=int(input("Enter your age: "))
monthly_income=int(input("Enter your monthly income: "))
if age<18:
    print(f"Since your age is :{age}, So you are Not eligible for holding a bank account")
elif age>=18 and monthly_income<15000:
    print("Eligible for basic savings account.")
elif age>=18 and (monthly_income>=15000 and monthly_income<=50000):
    print("Eligible for savings + salary account.")
elif age>=18 and (monthly_income>50000):
    print("Eligible for premium account.")
```

```python
#Use Case 1: Banking Eligibility Check alternate
age=int(input("Enter your age: "))
if age<18:
    print(f"Since your age is :{age}, So you are Not eligible for holding a bank account")
elif age>=18:
    monthly_income=int(input("Enter your monthly income: "))
    if monthly_income<15000:
        print("Eligible for basic savings account.")
    elif (monthly_income>=15000 and monthly_income<=50000):
        print("Eligible for savings + salary account.")
    elif monthly_income>50000:
        print("Eligible for premium account.")
```

```python
#Use Case 2: Sum of Even and Odd Numbers
number=int(input("Enter a number: "))
even_list=[]
odd_list=[]
even_sum=0
odd_sum=0
print(f"Sum of Odd & Even numbers till :{number}")
for i in range(1,number+1):
    if i>1 and i%2==0:
        even_list.append(i)
        even_sum+=i
    else:
        odd_list.append(i)
        odd_sum+=i
print(f"Sum of all even numbers from {even_list}: {even_sum}")
print(f"Sum of all odd numbers from  {odd_list}: {odd_sum}")
```
