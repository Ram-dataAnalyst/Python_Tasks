
## M. Function Based Programming

### Usecase 1: Food Delivery Discount (Convert this program to Exception Handling & FBP)


```python
def calculate_discount(amount):
    if amount >= 1000:
        discount = amount * 0.20
        return min(discount, 100)
    elif amount >= 600:
        return amount * 0.10
    else:
        return 0


def main():
    try:
        amount = float(input("Enter food order amount: ₹"))

        if amount <= 0:
            raise ValueError("Amount must be greater than 0.")

        discount = calculate_discount(amount)
        final_amount = amount - discount

        print(f"Total amount : ₹ {amount}")
        print("Discount: ₹", discount)
        print("Final Amount: ₹", final_amount)

    except ValueError as e:
        print("Error:", e)


main()
```

### Use Case 2: Salary Processing System (HR / Payroll Automation)
Demonstrates input/output functions, default arguments, arbitrary args, returning values, and composing multiple functions.
What Problem It Solves:
Companies calculate employee salary differently depending on bonus percentage, incentives, PF, tax, etc.
How Function-Based Programming Helps:
We create multiple small reusable functions (bonus calc, tax calc, net salary calc)
Then compose them together.


```python
def calculate_bonus(basic_salary, bonus_percentage=10):
    return basic_salary * bonus_percentage / 100


def calculate_incentive(*incentives):
    return sum(incentives)


def calculate_pf(basic_salary, pf_percentage=12):
    return basic_salary * pf_percentage / 100



def calculate_net_salary(basic_salary, bonus, incentive, pf):
    gross_salary = basic_salary + bonus + incentive
    return gross_salary - pf


def display_salary(employee_name, basic_salary, bonus,
                   incentive, pf, net_salary):

    print("Employee      :", employee_name)
    print("Basic Salary  : ₹", basic_salary)
    print("Bonus         : ₹", bonus)
    print("Incentive     : ₹", incentive)
    print("PF            : ₹", pf)
    print("Net Salary    : ₹", net_salary)


# Input
employee_name = input("Enter employee name: ")
basic_salary = float(input("Enter basic salary: ₹"))

bonus_percentage = float(
    input("Enter bonus percentage (default 10): ") or 10
)

# Function calls
bonus = calculate_bonus(basic_salary, bonus_percentage)

incentive = calculate_incentive(2000, 1000, 1500)

gross_salary = basic_salary + bonus + incentive

pf = calculate_pf(basic_salary)


net_salary = calculate_net_salary(
    basic_salary,
    bonus,
    incentive,
    pf,
    )

# Output
display_salary(
    employee_name,
    basic_salary,
    bonus,
    incentive,
    pf,
    net_salary
)
```

### Usecase3: Number Utility Tool
Create functions:
is_even(num) → returns True/False
find_max(*numbers) → returns highest number from arguments
perform_operation(num, operation) →
If operation = "square" → return num*num
If operation = "cube" → return numnumnum
If unknown → return "Invalid operation"



```python
def is_even(numb):
    even_list=[]
    odd_list=[]
    for num in numb:
        if num % 2 == 0:
            even_list.append(num)
        else:
            odd_list.append(num)
    return even_list,odd_list

def max_num(numb):
    return max(numb)

def perform_operation(num, operation):
    if operation == "square":
        return num * num

    elif operation == "cube":
        return num * num * num

    else:
        return "Invalid operation"

numbers=input("Enter the table numbers (numbers seperated by ,): ")
values=[]
for i in numbers.split(","):
    i=i.strip()
    values.append(int(i))
print(f"Given Numbers are:{values}")
even,odd = is_even(values)
print("Even numbers:", even)
print("Odd numbers:", odd)
print("Max Number:", max_num(values))

num = int(input("Enter number to find (square/cube): "))
operation = input("Enter operation (square/cube): ")
result = perform_operation(num, operation)
print(f"{operation} of {num} is : {result}")
```


### Use Case 5: Invoice Generator (**kwargs)

Create a function generate_invoice(**products).
Each key is a product name and each value is its price.
Print all product names with prices and also print the total amount.


```python
def generate_invoice(**products):
    val=0
    for key,values in products.items():
        print(f"{key} : {values}")
        val=val+values
    return val

products = {}

n = int(input("How many products? "))
print("Invoice")
for i in range(n):
    product = input("Enter product name: ")
    price = float(input("Enter price: "))

    products[product] = price

print(f"Total : {round(generate_invoice(**products),2)}")
```
