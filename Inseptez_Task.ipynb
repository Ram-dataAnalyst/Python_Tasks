Expected output after fixing:

```python
#Use case1
employee_name=input("Enter your employee name: ")
base_salary=float(input("Enter your base salary: "))
hra_percent=int(input("Enter your hra percent: "))
bonus_amount=float(input("Enter your bonus amount: "))
HRA=base_salary*hra_percent/100
Total=base_salary+HRA+bonus_amount
print(f"Employee Name: {employee_name}")
print(f"Base Salary: {base_salary}")
print(f"HRA @ {hra_percent}% is {HRA}")
print(f"Bonus: {bonus_amount}")
print(f"Total Salary: {Total}")
```

```python
#Use Case 2: Student Result Classification
mark=input("Enter your mark: ")
try:
    new_mark=float(mark)
    if new_mark>=90:
        print(f"{new_mark} is Outstanding")
    elif new_mark>=75:
        print(f"{new_mark} is Excellent")
    elif new_mark>=50:
        print(f"{new_mark} is Pass")
    else:
        print(f"{new_mark} is Fail")
except ValueError:
    print(f"{mark} is Non-numeric")
```

### Use Case 3 (Bug Fixing): Logical and Comparison Operator Errors
The following code should determine voting eligibility, but it contains operator mistakes. Fix it.
Incorrect code:
age = input("Enter age: ")
 citizen = input("Are you an Indian citizen? (yes/no)")
if age > "18" and citizen = "yes":
 print("Eligible to vote")
 else:
 print("Not eligible")
Expected behavior:
Convert age to integer before comparison.


Only print "Eligible to vote" if age is 18 or above AND citizen input is "yes" (case-insensitive).


```python
#Use Case 1: Internet Data Usage Calculator
total_data=int(input("Enter your total data: "))
data_used=int(input("Enter your data used: "))
remaining_data = total_data - data_used
usage_per=(data_used / total_data)*100
print(f"Total data: {total_data}")
print(f"Data used: {data_used}")
print(f"Remaining data: {remaining_data}")
if usage_per>=80:
    print(f"You used {usage_per}%, Warning: High usage, consider upgrading your plan.")
else:
    print("Normal Usage")
```

```python
#Use Case 2: Shopping Discount Calculation
original_price=float(input("Enter original price: "))
discount_percent=float(input("Enter discount %: "))
discount_amount=(original_price*discount_percent)/100
Final_price=original_price-discount_amount
print(f"Original price: {original_price}")
print(f"Discount percent: {discount_percent}")
print(f"Final price: {Final_price}")
```

```python
#Use Case 3 (Bug Fixing): Logical and Comparison Operator Errors
#age = input("Enter age: ") need to update the input str to int
age = int(input("Enter age: "))
citizen = input("Are you an Indian citizen? (yes/no)")
if age > 18 and citizen =="yes": #age 18 is updated to INT, & equals to ==
    print("Eligible to vote")
else:
    print("Not eligible")
```
